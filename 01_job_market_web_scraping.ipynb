```python
# import lib

import re
import pandas as pd
import numpy as np
```

```python
# load the dataset

df = pd.read_excel("rawdata/job_listing.xlsx")

df.head()
```

```python
# check the dataset

print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# check data types and missing values

df.info()
```

```python
# check duplicate rows

df.duplicated().sum()
```

```python
# remove duplicate rows

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# check missing values

df.isnull().sum()
```

```python
# check column names

print(df.columns.tolist())
```

```python
# check salary values

df["salary"].head(20)
```

```python
# check data types and missing values

df.info()
```

```python
# check missing values

missing_values = df.isnull().sum()

missing_values
```

```python
# handle missing values

df["company"] = df["company"].fillna("Unknown")
df["experience"] = df["experience"].fillna("Not specified")
df["location"] = df["location"].fillna("Unknown")
df["rating"] = df["rating"].fillna(0)
df["reviews"] = df["reviews"].fillna(0)
df["key_skills"] = df["key_skills"].fillna("Not specified")
df["company_link"] = df["company_link"].fillna("Not available")
```

```python
# verify missing values

df.isnull().sum()
```

```python
# check salary values

df["salary"].value_counts().head(20)
```

```python
# extract minimum and maximum salary

def extract_salary(salary):
    salary = str(salary).lower().strip()

    if "not disclosed" in salary:
        return pd.Series([np.nan, np.nan])

    numbers = re.findall(r"\d+(?:\.\d+)?", salary)

    if len(numbers) >= 2:
        return pd.Series([float(numbers[0]), float(numbers[1])])
    elif len(numbers) == 1:
        return pd.Series([float(numbers[0]), float(numbers[0])])
    else:
        return pd.Series([np.nan, np.nan])


df[["min_salary_lpa", "max_salary_lpa"]] = df["salary"].apply(extract_salary)

df[["salary", "min_salary_lpa", "max_salary_lpa"]].head(20)
```

```python
# check salary conversion

print("Minimum salary available:", df["min_salary_lpa"].notna().sum())
print("Maximum salary available:", df["max_salary_lpa"].notna().sum())
print("Salary not disclosed:", df["min_salary_lpa"].isna().sum())
```

```python
# check experience values

df["experience"].value_counts().head(20)
```

```python
# extract minimum and maximum experience

def extract_experience(experience):
    experience = str(experience).lower().strip()

    numbers = re.findall(r"\d+(?:\.\d+)?", experience)

    if len(numbers) >= 2:
        return pd.Series([float(numbers[0]), float(numbers[1])])
    elif len(numbers) == 1:
        return pd.Series([float(numbers[0]), float(numbers[0])])
    else:
        return pd.Series([np.nan, np.nan])


df[["min_experience_years", "max_experience_years"]] = df["experience"].apply(
    extract_experience
)

df[["experience", "min_experience_years", "max_experience_years"]].head(20)
```

```python
# check experience conversion

print("Minimum experience available:", df["min_experience_years"].notna().sum())
print("Maximum experience available:", df["max_experience_years"].notna().sum())
print("Experience not available:", df["min_experience_years"].isna().sum())
```

```python
# check job roles

df["job_role"].value_counts().head(20)
```

```python
# standardize job titles

df["job_role"] = (
    df["job_role"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["job_role"].head(20)
```

```python
# check locations

df["location"].value_counts().head(20)
```

```python
# standardize locations

df["location"] = (
    df["location"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["location"].head(20)
```

```python
# check key skills

df["key_skills"].head(20)
```

```python
# split multiple skills

df["key_skills"] = df["key_skills"].str.split(",")

df["key_skills"].head(10)
```

```python
# clean individual skills

df["key_skills"] = df["key_skills"].apply(
    lambda skills: [
        skill.strip().title()
        for skill in skills
        if skill.strip()
    ] if isinstance(skills, list) else skills
)

df["key_skills"].head(10)
```

```python
# create separate skills dataset

skills_df = df[["job_id", "key_skills"]].explode("key_skills")

skills_df.head(10)
```

```python
# count most demanded skills

top_skills = skills_df["key_skills"].value_counts().head(20)

top_skills
```

```python
# check companies with most job postings

top_companies = df["company"].value_counts().head(20)

top_companies
```

```python
# save cleaned dataset

df.to_csv("cleaned_job_listing.csv", index=False)

print("Cleaned dataset saved successfully!")
```
